# Heart Disease Prediction: Exploratory Data Analysis & Stacking Model Experiments

**Objective**: Analyze the UCI Cleveland Heart Disease dataset, investigate feature correlations, class balance, and evaluate a heterogeneous Stacking Ensemble classifier.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid", palette="Blues_d")
print("Libraries loaded successfully.")
```

```python
# Load clean UCI Cleveland data
csv_path = os.path.abspath(os.path.join("..", "data", "raw", "heart_disease_uci.csv"))
df = pd.read_csv(csv_path)
print("Dataset Shape:", df.shape)
df.head()
```

```python
# Target Distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='target', palette=['#10b981', '#ef4444'])
plt.title('Target Distribution (0: No Disease, 1: Heart Disease)')
plt.xlabel('Target Class')
plt.ylabel('Patient Count')
plt.show()
print(df['target'].value_counts(normalize=True))
```

```python
# Correlation Heatmap across numeric features
numeric_cols = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak', 'target']
plt.figure(figsize=(8, 6))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='Blues', fmt='.2f')
plt.title('Correlation Heatmap - Hemodynamics and Stress Variables')
plt.show()
```
